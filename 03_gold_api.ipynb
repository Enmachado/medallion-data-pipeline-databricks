```python
from pyspark.sql.functions import count

df = spark.read.table("silver_posts")

df_gold = (
    df.groupBy("userId")
      .agg(count("*").alias("qtd_posts"))
)
```

```python
df_gold.write.format("delta") \
  .mode("overwrite") \
  .saveAsTable("gold_posts_por_usuario")
```

```python
%sql
SELECT * FROM gold_posts_por_usuario
```
